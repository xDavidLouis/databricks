```python
from pyspark.sql.functions import element_at,col,lower,split,regexp_extract


FOLDER_URL = "https://drive.google.com/drive/folders/1aFkRHHFHZB7IlmcpcjTp-Gl6eCxL_cWq"

df_files = (
    spark.read
    .format("binaryFile")
    .option("databricks.connection", "googledrive_con")
    .option("recursiveFileLookup", True)
    .load(FOLDER_URL)
    .select(
        col("_metadata.file_name").alias("name"),
        col("path"),
        lower(regexp_extract("_metadata.file_name", r"\.([^./]+)$", 1)).alias("type"),
    )
)

display(df_files)
```
